# 🚢 Titanic Interactive Dashboard

An interactive Streamlit dashboard for exploring the Titanic dataset with Realtime-Filtering.

**Technologies:** `streamlit`, `plotly`, `pandas`

In this notebook, the complete code for the dashboard is presented and each component is explained.

**Learning Ojectives:**

- Data loading and preprocessing
- Exploratory data analysis (EDA)
- Visualization techniques with Plotly
- Statistical insights from the dataset
- Interactive filtering concepts

````python
# Import required libraries
import pandas as pd
import plotly.express as px
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Set display options for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
````

## Data Loading and Preparation

First, we load the Titanic dataset and prepare it for analysis by mapping embarkation ports to their full names.

````python
# Load the dataset
df = pd.read_csv('titanic.csv')

# Display basic information about the dataset
print("Dataset shape:", df.shape)
print("\nDataset info:")
df.info()

# Show first few rows of the dataset
print("\nFirst 5 rows:")
df.head()
````

## Data Exploration and Understanding

Let's examine the key variables in our dataset to understand what we're working with.

````python
# Check for missing values
print("Missing values per column:")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])

# Basic statistics for numerical columns
print("\nBasic statistics for numerical columns:")
df.describe()
````

## Data Preprocessing

We'll map the embarkation port codes to their full names for better readability and create a new column for this mapping.

````python
# Map embarkation ports to full names
embark_map = {
    'S': 'Southampton',
    'C': 'Cherbourg',
    'Q': 'Queenstown'
}

# Create new column with full names
df['EmbarkedName'] = df['Embarked'].map(embark_map)

# Display the mapping
print("Embarkation port mapping:")
print(embark_map)

# Show some sample rows with the new column
print("\nSample data with embarkation names:")
df[['Embarked', 'EmbarkedName']].head(10)
````

## Exploratory Data Analysis (EDA)

Let's perform some initial analysis to understand the distribution of passengers across different categories.

````python
# Analyze passenger class distribution
print("Passenger Class Distribution:")
class_dist = df['Pclass'].value_counts().sort_index()
print(class_dist)

# Analyze gender distribution
print("\nGender Distribution:")
gender_dist = df['Sex'].value_counts()
print(gender_dist)

# Analyze survival statistics
print("\nSurvival Statistics:")
survival_rate = df['Survived'].mean()
print(f"Overall survival rate: {survival_rate:.2%}")

# Survival by class
print("\nSurvival rate by passenger class:")
survival_by_class = df.groupby('Pclass')['Survived'].mean()
print(survival_by_class)

# Survival by gender
print("\nSurvival rate by gender:")
survival_by_gender = df.groupby('Sex')['Survived'].mean()
print(survival_by_gender)
````

## Data Visualization

Now we'll create several visualizations to better understand the relationships in our data.

````python
# Survival rate by passenger class
class_stats = df.groupby('Pclass')['Survived'].agg(['mean', 'count']).reset_index()
class_stats.columns = ['Pclass', 'SurvivalRate', 'Passengers']
class_stats['SurvivalRate'] *= 100

# Create bar chart
fig = px.bar(
    class_stats,
    x='Pclass',
    y='SurvivalRate',
    color='Pclass',
    title='Survival Rate by Passenger Class',
    hover_data=['Passengers']
)

# Show the chart
fig.show()
````

````python
# Survival rate by gender
gender_stats = df.groupby('Sex')['Survived'].mean().reset_index()
gender_stats['Survived'] *= 100

# Create bar chart
fig = px.bar(
    gender_stats,
    x='Sex',
    y='Survived',
    color='Sex',
    title='Survival Rate by Gender'
)

# Show the chart
fig.show()
````

````python
# Age distribution with survival status
fig = px.histogram(
    df[df['Age'].notna()],
    x='Age',
    color='Survived',
    barmode='overlay',
    opacity=0.7,
    title='Age Distribution: Survived vs Died'
)

# Show the chart
fig.show()
````

````python
# Fare vs Age scatter plot
valid_data = df[df['Age'].notna() & df['Fare'].notna()]

fig = px.scatter(
    valid_data,
    x='Age',
    y='Fare',
    color='Pclass',
    symbol='Survived',
    hover_data=['Name', 'Sex'],
    title='Fare vs Age'
)

# Show the chart
fig.show()
````

````python
# Passenger distribution by class (pie chart)
class_dist = df['Pclass'].value_counts().sort_index().reset_index()
class_dist.columns = ['Pclass', 'Passengers']

fig = px.pie(
    class_dist,
    values='Passengers',
    names='Pclass',
    title='Passenger Distribution by Class'
)

# Show the chart
fig.show()
````

## Key Insights from Analysis

Based on our analysis, we can draw several important conclusions:

1. **Class Impact**: Passengers in higher classes (1st class) had significantly higher survival rates than those in lower classes.
2. **Gender Effect**: Women had much higher survival rates than men, reflecting the "women and children first" protocol.
3. **Age Distribution**: The age distribution shows different patterns between survivors and non-survivors.
4. **Fare Correlation**: There's a positive correlation between fare paid and class, with higher fares generally associated with better survival rates.

These insights help us understand the factors that influenced survival during the Titanic disaster.

## Summary Statistics

Let's calculate and display some key statistics from our dataset:

````python
# Overall statistics
total_passengers = len(df)
overall_survival_rate = df['Survived'].mean()
avg_fare = df['Fare'].mean()
avg_age = df['Age'].mean()

print("Overall Statistics:")
print(f"Total passengers: {total_passengers}")
print(f"Overall survival rate: {overall_survival_rate:.2%}")
print(f"Average fare: ${avg_fare:.2f}")
print(f"Average age: {avg_age:.1f} years")
````
